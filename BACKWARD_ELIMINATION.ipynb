```
import pandas as pd
```

```
df=pd.read_excel('/content/book 1.xlsx')
df
```

```
x=df.iloc[:,0]
y=df.iloc[:,-1]
x
y
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=0)
x_train
y_train
x_test
y_test
```

```
x_pred=run.predict(x_train.values.reshape(-1,1))
y_pred=run.predict(x_test.values.reshape(-1,1))
```

```
import matplotlib.pyplot as plt
plt.scatter(x_train,y_train,color='red')
plt.plot(x_train,x_pred,color='blue')
plt.title('empolyee details')
plt.show()
```

```
import pandas as pd
df1=pd.read_excel('/content/book 2.xlsx')
df1
```

```
x=df1.iloc[:,:-1]
y=df1.iloc[:,4]
```

```
from sklearn.preprocessing import LabelEncoder
earn=LabelEncoder()
x.iloc[:,3]=earn.fit_transform(x.iloc[:,3])
x.iloc[:,3]
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=0)
x_train
```

```
from sklearn.linear_model import LinearRegression
turn=LinearRegression()
turn.fit(x_train,y_train)
```

```
y_pred=turn.predict(x_test)
y_pred
```

```
train_score=turn.score(x_train,y_train)
test_score=turn.score(x_test,y_test)
train_score
```
